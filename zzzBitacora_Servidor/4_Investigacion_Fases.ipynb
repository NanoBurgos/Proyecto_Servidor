## **FASE 0 — Mentalidad (antes de tocar nada)**

Aprendé esto primero:

Seguridad ≠ instalar un antivirus

Seguridad = capas (defense in depth)

El mayor riesgo casi siempre es:

- malas configuraciones
- contraseñas débiles
- servicios expuestos innecesariamente

📌 Conceptos clave:

- Superficie de ataque
- Principio de mínimo privilegio
- Zero trust (no confiar en nada por defecto)

## **Probablemente la parte más importante de aprender seguridad informática o administración de servidores. **

Mucha gente empieza instalando herramientas, firewalls o “antivirus”, pero sin esta mentalidad termina dejando agujeros enormes.

**Seguridad ≠ instalar un antivirus**

Muchos creen:

            “Instalé un antivirus → ya estoy seguro”

Pero en servidores Linux, infraestructura, APIs o redes, la seguridad real no depende de una sola herramienta.

**Un antivirus es apenas una capa más.**

- Un servidor puede ser comprometido aunque tenga antivirus si:

  - SSH está mal configurado
  - hay contraseñas débiles
  - puertos innecesarios están abiertos
  - Docker expone servicios al exterior
  - una API no tiene autenticación
  - un usuario tiene permisos excesivos

- La seguridad moderna se basa más en:

  - configuración correcta
  - aislamiento
  - permisos mínimos
  - monitoreo
  - actualizaciones
  - segmentación
  - autenticación fuerte
