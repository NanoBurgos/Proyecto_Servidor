## **FASE 0 — Mentalidad (antes de tocar nada)**

Aprendé esto primero:

Seguridad ≠ instalar un antivirus

Seguridad = capas (defense in depth)

El mayor riesgo casi siempre es:

- malas configuraciones
- contraseñas débiles
- servicios expuestos innecesariamente

📌 Conceptos clave:

- Superficie de ataque
- Principio de mínimo privilegio
- Zero trust (no confiar en nada por defecto)

## **Probablemente la parte más importante de aprender seguridad informática o administración de servidores.**

Mucha gente empieza instalando herramientas, firewalls o “antivirus”, pero sin esta mentalidad termina dejando agujeros enormes.

**Seguridad ≠ instalar un antivirus**

Muchos creen:

            “Instalé un antivirus → ya estoy seguro”

Pero en servidores Linux, infraestructura, APIs o redes, la seguridad real no depende de una sola herramienta.

**Un antivirus es apenas una capa más.**

- Un servidor puede ser comprometido aunque tenga antivirus si:

  - SSH está mal configurado
  - hay contraseñas débiles
  - puertos innecesarios están abiertos
  - Docker expone servicios al exterior
  - una API no tiene autenticación
  - un usuario tiene permisos excesivos

- La seguridad moderna se basa más en:

  - configuración correcta
  - aislamiento
  - permisos mínimos
  - monitoreo
  - actualizaciones
  - segmentación
  - autenticación fuerte

## **Seguridad = capas (Defense in Depth)**

“Defense in depth” significa:

- si una capa falla, otra sigue protegiendo.

Imaginá un castillo medieval:

- murallas
- foso
- puertas
- guardias
- torres
- puertas internas

No dependían de una sola defensa.

En servidores pasa igual.

**Ejemplo real**

Supongamos que tenés un servidor Ubuntu con:

- SSH
- Docker
- una API FastAPI
- base de datos PostgreSQL

Las capas podrían ser:

|Capa	|Protección|
|:-:|:-:|
|Firewall	|bloquea puertos|
|SSH con claves	|evita contraseñas|
|Fail2Ban	|bloquea intentos bruteforce|
|Docker networks	|aísla contenedores|
|Usuarios sin sudo	|limita daños|
|HTTPS	|protege tráfico|
|Backups	|recuperación|
|Logs	|detección|
|Actualizaciones	|corrigen vulnerabilidades|

Si una falla, las demás siguen ayudando.

## **El mayor riesgo casi siempre NO es un hacker ultra avanzado**

En la práctica, la mayoría de ataques exitosos pasan por errores simples.

## **1. Malas configuraciones**

Esto es extremadamente común.

**Ejemplos:**

- Base de datos expuesta a internet
- Docker con puertos abiertos innecesarios
- Permisos 777
- .env público
- Panel admin sin autenticación
- Servicios escuchando en 0.0.0.0
- Desactivar firewall “porque molesta”

## **2. Contraseñas débiles**

- Todavía hoy muchísimos ataques funcionan por:

  - admin123
  - reutilización de contraseñas
  - claves filtradas
  - acceso SSH con password

- Por eso se usa:

  - claves SSH
  - MFA/2FA
  - password managers
  - rotación de credenciales

# **3. Servicios expuestos innecesariamente**

Cada puerto abierto es una posible puerta de entrada.

Por ejemplo:

|Servicio	|¿Debe estar expuesto?|
|:-:|:-:|
|SSH	|quizá sí|
|PostgreSQL	|normalmente NO|
|Redis	|casi nunca|
|Docker daemon	|NO|
|Paneles admin	|preferiblemente detrás de VPN|

**Muchos ataques automatizados escanean internet buscando:**

- Redis abiertos
- MongoDB abiertos
- phpMyAdmin
- Jenkins
- Elasticsearch
- Puertos Docker

## **Concepto: Superficie de ataque**

La superficie de ataque es:

- todo lo que un atacante puede intentar usar para entrar.

Mientras más cosas expuestas tengas, mayor superficie.

**Ejemplo**

Servidor con:

- SSH
- FTP
- Apache
- MySQL público
- phpMyAdmin
- Jenkins
- puerto Docker abierto
- panel admin

**Tiene MUCHÍSIMA superficie de ataque.**

## **Reducir superficie de ataque**

La idea es:

- exponer lo mínimo posible.

**Entonces:**

- cerrar puertos
- eliminar servicios innecesarios
- desinstalar software que no uses
- usar firewall
- ocultar paneles internos
- limitar accesos por IP

**Menos cosas abiertas = menos oportunidades de ataque.**

## **Principio de mínimo privilegio**

Este es uno de los conceptos más importantes.

**Significa:**

- Cada usuario o servicio debe tener SOLO los permisos estrictamente necesarios.

No más.

## **Mal ejemplo**

Tu API usa:

            root

para conectarse a todo.

**Si la hackean:**

- → control total del servidor.

## **Buen ejemplo**

La API usa un usuario:

            api_user

- que solo puede:

  - leer/escribir cierta base
  - acceder a cierta carpeta
  - sin sudo

Entonces, aunque comprometan la API, el daño queda limitado.

## **Otro ejemplo cotidiano**

No usar:

            sudo

todo el tiempo.

- En Linux:

  - root = poder absoluto
  - cualquier error puede destruir el sistema

## **Zero Trust**

Este concepto cambió muchísimo la seguridad moderna.

**La idea es:**

- NO confiar en nada por defecto.

- Ni siquiera dentro de tu propia red.

**Antes**

Modelo antiguo:

            si está dentro de la red → confiable

**Ahora**

Modelo Zero Trust:

**verificar SIEMPRE**

Aunque esté dentro de:

- tu LAN
- tu VPN
- tu infraestructura

##  **Ejemplos prácticos**

**❌ Malo**

“Está dentro de Docker network, entonces confío”

**✅ Mejor**
- autenticación interna
- tokens
- permisos
- segmentación
- cifrado interno

**Otro ejemplo**

Aunque tengas VPN:

- seguís usando MFA
- seguís validando permisos
- seguís auditando accesos

## **Mentalidad correcta de seguridad**

La seguridad real no es:

            instalar herramientas mágicas

**Es pensar constantemente:**

- ¿esto realmente necesita estar expuesto?
- ¿qué pasa si comprometen este servicio?
- ¿qué permisos tiene?
- ¿cómo limito el daño?
- ¿qué estoy asumiendo incorrectamente?
- ¿qué capa me protege si otra falla?

## **Regla importante**

- En seguridad:

    - no existe “100% seguro”.

- El objetivo real es:

    - reducir riesgos
    - dificultar ataques
    - detectar intrusiones
    - limitar daños
    - recuperarse rápido
