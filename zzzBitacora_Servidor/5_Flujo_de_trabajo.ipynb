Usaremos un flujo profesional, pero sin añadir complejidad innecesaria.

## **Flujo de ramas que utilizaremos**

```bash
  main
    └── develop
          ├── feature/application-factory
          ├── feature/configuracion
          ├── feature/base-datos
          ├── feature/api-sincronizacion
          └── feature/docker
``` 

### **main**

Representará la versión estable del servidor.

Todo lo que esté en main debería:

- funcionar correctamente;
- haber sido probado;
- estar listo para desplegarse;
- no contener cambios incompletos.

No trabajaremos directamente sobre main, salvo esta preparación inicial.

### **develop**

Será la rama donde integraremos las funcionalidades terminadas.

Puede contener trabajo más reciente que main, pero siempre debe mantenerse funcional.

### **feature/nombre**

Cada funcionalidad importante tendrá su propia rama temporal.

**Cuando una funcionalidad termine y funcione, la fusionaremos con develop y podremos borrar su rama.**

## **Cada etapa de desarrollo debe tener este ciclo**

```bash
1. Diseñar
        ↓
2. Implementar
        ↓
3. Probar
        ↓
4. Commit
        ↓
5. Continuar
```
